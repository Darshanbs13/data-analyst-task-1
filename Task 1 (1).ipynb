```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df.columns
```

```python
for col in ["customer_name", "gender", "country", "email"]:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
```

```python
gender_map = {
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
}

df["gender"] = df["gender"].str.lower().map(gender_map)
```

```python
df["gender"].value_counts()
```

```python
df["country"] = df["country"].str.title()
```

```python
df["country"].value_counts()
```

```python
df["join_date"] = pd.to_datetime(
    df["join_date"],
    dayfirst=True,
    errors="coerce"
)
```

```python
df["join_date"].head()
```

```python
df["join_date"] = df["join_date"].dt.strftime("%d-%m-%Y")
```

```python
df["age"] = pd.to_numeric(df["age"], errors="coerce")
```

```python
df["age"].dtype
```

```python
df["age"].median()
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df["age"].isnull().sum()
```

```python
df["email"] = df["email"].fillna("Not Provided")
```

```python
df["email"].isnull().sum()
```

```python
df["annual_income"] = pd.to_numeric(
    df["annual_income"],
    errors="coerce"
)
```

```python
duplicate_columns = [
    "customer_name",
    "gender",
    "country",
    "join_date",
    "age",
    "email",
    "annual_income"
]

df = df.drop_duplicates(
    subset=duplicate_columns,
    keep="first"
).reset_index(drop=True)
```
